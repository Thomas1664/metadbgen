```python
import pandas as pd
from src import database
```

```python
df = pd.read_csv('data/cox2_th/output/cox2_th-meta-clean.tsv', sep='\t')
df
```

```python
df['organism_name_sanitized'] = df['organism_name']
df['organism_name_sanitized'] = df['organism_name_sanitized'].apply(database.sanitize_org_name)
df.index = df['accession']
is_on_mask = df['organism_name'] != df['organism_name_sanitized']
non_on_seqs = df[~is_on_mask]
non_on_species = non_on_seqs['organism_name'].unique()
# ON species that don't have a non-ON species name
# e.g. Globisporangium aff. irregulare while there is no Globisporangium irregulare
on_native_mask = is_on_mask & ~df['organism_name_sanitized'].isin(non_on_species)
on_native = df[on_native_mask]
trimmed = df.filter([*non_on_seqs.index, *on_native.index], axis='index')
trimmed
```

```python
on_remaining = trimmed[trimmed['organism_name'] != trimmed['organism_name_sanitized']]
on_remaining
```
