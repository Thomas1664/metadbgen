### Outline

1. Allow `efetch_by_ids()` to take in a callback instead of saving to a file by default
1. Fetch UIDs from primer BLAST output to get genbank records


```python
df = pd.read_csv('data/cox2_th_cutadapt/primer-blast-outgroups.csv')
uids = df['uid'].astype(int).astype(str)
client = request.HttpClient()
with open('data/cox2_th_cutadapt/outgroups.gb', 'w') as f:
    await client.efetch_by_ids('nuccore', uids.to_list(), lambda x: f.write(x), ret_type='gb')
```

```python
meta_df = pd.read_csv('data/cox2_th_cutadapt/db-clust.csv', sep='\t')
tax_ids = meta_df['tax_id_sanitized'].astype(int).astype(str).tolist()
client = request.HttpClient()
await taxonomy.fetch_by_tax_ids(tax_ids, client, 'data/cox2_th_cutadapt/outgroup-taxonomy.csv')
```
