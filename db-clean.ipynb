### Trim sequences to COX2 region

````python
from Bio import SeqIO
````

````python
def trim_seq(cox2_start: str, cox2_end: str, seq: str) -> str:
    start = int(str(cox2_start).lstrip('<'))
    end = int(str(cox2_end).lstrip('>'))
    assert(start < end)
    return seq[start:end]
````

````python
filtered['sequence'] = filtered.apply(lambda row: trim_seq(row['cox2_start'], row['cox2_end'], row['sequence']), axis=1)
filtered
````

### Remove short (<200bp) sequences

````python
trimmed = filtered[filtered['sequence'].str.len() >= 200]
trimmed
````

### Remove Sequences with unknown species (i.e. 'sp.')

````python
trimmed = trimmed[~trimmed['organism_name'].str.contains('sp.', case=False, regex=False)]
trimmed
````

### Remove sequences from uncultured samples

````python
trimmed[trimmed['organism_name'].str.contains('uncultured', case=False)]
````

````python
trimmed = trimmed[~trimmed['organism_name'].str.contains('uncultured', case=False)]
trimmed
````

### Remove (nom. inval.) species

````python
trimmed = trimmed[~trimmed['title'].str.contains('nom. inval.', case=False, regex=False)]
trimmed = trimmed[~trimmed['organism_name'].str.contains('nom. inval.', case=False, regex=False)]
````

````python
trimmed.value_counts('tax_id')
````

````python
trimmed[(trimmed['cox2_start'] != '<0') & (trimmed['sequence_length'] > 280) & (trimmed['sequence'].str.len() < 290)]
````

````python
def sanitize_org_name(org_name: str) -> str:
    org_name_arr = org_name.split(' ')
    to_remove = ['x', 'var.', 'var', 'aff.', 'aff', 'cf.', 'cf']
    for s in to_remove:
        try:
            org_name_arr.remove(s)
        except ValueError:
            pass
    return ' '.join(org_name_arr[:2])
````

````python
trimmed['tax_id_sanitized'] = trimmed['tax_id']
trimmed['organism_name_sanitized'] = trimmed['organism_name'].apply(sanitize_org_name)
def fix_tax_id(row: pd.Series):
    if row['organism_name_sanitized'] == row['organism_name']:
        return row['tax_id']
    if row['organism_name_sanitized'] in trimmed['organism_name'].unique():
        new_tax_id = trimmed.loc[trimmed['organism_name'] == row['organism_name_sanitized']]['tax_id'].iloc[0]
        return new_tax_id
    return row['tax_id']

trimmed['tax_id_sanitized'] = trimmed.apply(fix_tax_id, axis=1)
````

````python
species_groups = trimmed.groupby('organism_name_sanitized')
for _, species in species_groups:
    # accessions of refseq sequences contain an underscore
    # see https://www.ncbi.nlm.nih.gov/refseq/about/
    refseq = species[species['accession'].str.contains('_')]
    if not refseq.empty:
        non_refseq = species[~species['accession'].str.contains('_')]
        if not non_refseq.empty:
            trimmed.drop(trimmed[trimmed['accession'].isin(non_refseq['accession'])].index, inplace=True)
trimmed
````

````python
trimmed.to_csv('data/cox2-db-filtered.csv', sep='\t', index=False)
````
