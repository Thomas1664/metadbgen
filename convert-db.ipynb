```python
from Bio import SeqIO
import re

# Input and output files
input_fasta = "cox2_oomycete_refDB_uparse_v10_Sappetal.txt"  # Replace with your actual input file
output_fasta = "melanie_qiime_sequences.fasta"
output_tsv = "melanie_qiime_taxonomy.tsv"

# Open files for writing
with open(output_fasta, "w") as fasta_out, open(output_tsv, "w") as tsv_out:
    for record in SeqIO.parse(input_fasta, "fasta"):
        # Extract accession number (everything before ";")
        acc_match = re.match(r"^([^;]+)", record.description)  
        if not acc_match:
            continue  # Skip if no match found
        
        accession = acc_match.group(1)  # Get accession number

        # Extract taxonomy (after "tax=")
        tax_match = re.search(r"tax=([^\n]+)", record.description)  
        if not tax_match:
            continue  # Skip if no taxonomy found
        
        # Clean up taxonomy format for QIIME
        taxonomy = tax_match.group(1).replace(",", "; ")  # Ensure spaces for readability
        taxonomy = taxonomy.replace(":", "__")
        # Write to TSV (Accession <TAB> Taxonomy)
        tsv_out.write(f"{accession}\t{taxonomy}\n")
        
        # Write to FASTA with just the accession number in the header
        fasta_out.write(f">{accession}\n{str(record.seq)}\n")

print("Conversion completed! Files saved as:")
print(f"- {output_fasta}")
print(f"- {output_tsv}")
```
