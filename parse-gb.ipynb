```python
import pandas as pd
import json
```

```python
df = pd.read_csv('data/seq-meta.csv', index_col='accession')
df['metadata'] = df['metadata'].apply(json.loads)

df
```

```python

new_df = pd.DataFrame(df['metadata'].tolist(), index=df.index)
new_df
```

```python
# Remove single-element lists
def unwrap_single_element(x):
    if isinstance(x, list) and len(x) == 1:
        return x[0]
    return x

# Apply it across the whole DataFrame
new_df = new_df.applymap(unwrap_single_element)
new_df
```
